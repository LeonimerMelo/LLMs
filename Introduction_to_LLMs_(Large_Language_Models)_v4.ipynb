## LLMs - Large Language Models

Os modelos de linguagem — frequentemente conhecidos pela sigla LLMs (Large Language Models, ou Grandes Modelos de Linguagem) — são a base de poderosas aplicações de Inteligência Artificial, como chatbots conversacionais, assistentes de IA e outras ferramentas inteligentes capazes de gerar textos e conteúdos.

Este notebook apresenta uma compreensão básica e concisa sobre os LLMs, seguida de três exemplos introdutórios com código, que demonstram como utilizá-los por meio de alguns frameworks bastante conhecidos, como Hugging Face, Ollama e LangChain.

Os Grandes Modelos de Linguagem (LLMs) são modelos de Inteligência Artificial treinados com grandes quantidades de dados textuais para aprender padrões da linguagem. Eles conseguem interpretar perguntas, produzir respostas, resumir documentos, traduzir textos, escrever códigos de programação e realizar diversas outras tarefas relacionadas à linguagem.

### Principais ferramentas

1. **Hugging Face**. É uma plataforma e um ecossistema de ferramentas que permite acessar, utilizar, adaptar e treinar modelos de IA. A biblioteca Transformers facilita o uso de modelos pré-treinados em Python.

1. **Ollama**. É uma ferramenta que facilita a execução de modelos de linguagem no computador local, permitindo experimentar LLMs sem depender necessariamente de um serviço de IA hospedado na nuvem.

1. **LangChain**. É um framework que ajuda a desenvolver aplicações com modelos de linguagem, integrando-os a documentos, bancos de dados, ferramentas externas e fluxos de processamento.

## O que são Modelos de Linguagem?

Em sua essência, os modelos de linguagem são sistemas de Processamento de Linguagem Natural (PLN), ou Natural Language Processing (NLP), capazes de prever a próxima palavra de uma sequência, após terem aprendido padrões complexos da linguagem humana por meio da exposição a grandes conjuntos de dados textuais, que podem conter milhares, milhões ou até bilhões de documentos.

Naturalmente, os modelos de linguagem, especialmente os Grandes Modelos de Linguagem (LLMs — Large Language Models), evoluíram significativamente e passaram a realizar diversas tarefas complexas de compreensão e processamento da linguagem, muito além da simples geração da próxima palavra. Eles podem responder a perguntas, resumir ou traduzir textos e até mesmo extrair informações relevantes ou classificar conteúdos.

As aplicações baseadas em LLMs existem em diferentes modalidades:

- Modelos acessados por API: modelos como o GPT-4, da OpenAI, e o Claude, da Anthropic, podem ser acessados por meio de serviços disponibilizados na internet, incluindo sites e aplicativos. O ChatGPT, especificamente, é uma aplicação de IA, e não apenas o nome de um modelo.
- Modelos locais: modelos como Llama, Mistral e Qwen normalmente são executados em computadores pessoais ou em servidores próprios, permitindo maior controle sobre o ambiente de execução e os dados processados.
- Aplicações híbridas: utilizam modelos de linguagem integrados a outras ferramentas, bibliotecas e serviços. Frameworks como o LangChain facilitam a integração dos LLMs com bancos de dados, documentos, APIs e outras aplicações. O LangChain não é, por si só, um modelo de linguagem.

## Arquitetura dos LLMs

A maioria dos Grandes Modelos de Linguagem (LLMs — Large Language Models) modernos utiliza uma arquitetura de rede neural profunda chamada **Transformer**, apresentada no artigo *Attention Is All You Need*, de 2017.

Essa arquitetura permite que o modelo aprenda relações entre palavras e tokens, identifique dependências entre diferentes partes de um texto e gere respostas considerando o contexto.

O artigo científico "*Attention Is All You Need*" foi publicado em junho de 2017 por pesquisadores do Google e mudou para sempre a história da Inteligência Artificial

###Principais Informações do Artigo

### O que é o Transformer

Um Transformer é uma arquitetura de rede neural que processa sequências de texto usando atenção para descobrir quais palavras (tokens) são mais importantes umas para as outras. Modelos como GPT são "decoder-only": eles leem o texto da esquerda para a direita e tentam prever o próximo token, repetidamente .

### Como Funciona o Transformer

• Mecanismo de Autoatenção (**Self-Attention**): Permite que o modelo olhe para todas as palavras de uma frase ao mesmo tempo, entendendo o contexto e a relação entre elas, mesmo que estejam muito distantes no texto. Enquanto o modelo processa a frase, cada token "olha" para os outros tokens para entender o contexto. Na frase "O banco estava cheio", a palavra "banco" (instituição financeira) pode dar mais atenção a "cheio" do que se a frase fosse "O banco da praça estava cheio".

• **Tokenização**: O texto é quebrado em subpalavras. "Inteligência" pode virar "Intelig" + "ência". Isso permite que o modelo lide com palavras raras ou novas (como gírias ou termos técnicos).

• **Predição autorregressiva**: Modelos GPT são treinados para prever o próximo token. Eles não "sabem" fatos; eles aprenderam, com trilhões de palavras, qual sequência de tokens é estatisticamente mais plausível.

• **Paralelização**: Como não precisa ler o texto em fila indiana, ele aproveita o poder de processamento das placas de vídeo (GPUs), acelerando drasticamente o treinamento.

• Estrutura de Codificador-Decodificador (**Encoder-Decoder**): Composta por camadas empilhadas que usam "atenção de cabeças múltiplas" (multi-head attention) para capturar diferentes nuances da linguagem.

###Impacto no Mundo da IA

• Transformer é a base dos LLMs: É a fundação tecnológica por trás de praticamente todos os Grandes Modelos de Linguagem modernos, como o ChatGPT, Claude, Gemini e Llama.

• Desempenho Original: No artigo, o modelo foi testado em tradução automática (inglês para alemão e inglês para francês), superando os recordes anteriores com um custo de treinamento muito menor.

## Arquitetura Transformer
A arquitetura Transformer proposta em 2017 é baseada em uma estrutura clássica de Codificador-Decodificador (Encoder-Decoder), mas com uma diferença crucial: ela substitui completamente as redes recorrentes por mecanismos de atenção.
Abaixo está o funcionamento detalhado dos blocos que compõem essa arquitetura:
```python
                       [ ENTRADA DO TEXTO ]
                                │
                  Embedding + Positional Encoding
                                │
                                ▼
         ┌──────────────────────────────────────────────┐
         │             BLOCO ENCODER (x6)               │
         │  ┌────────────────────────────────────────┐  │
         │  │ Multi-Head Self-Attention              │  │
         │  └───────────────────┬────────────────────┘  │
         │                      ▼                       │
         │  ┌────────────────────────────────────────┐  │
         │  │ Feed Forward Neural Network            │  │
         │  └───────────────────┬────────────────────┘  │
         └──────────────────────┼───────────────────────┘
                                │ (Representação Contextualizada)
                                ▼
         ┌──────────────────────────────────────────────┐
         │             BLOCO DECODER (x6)               │
         │  ┌────────────────────────────────────────┐  │
         │  │ Masked Multi-Head Self-Attention       │  │
         │  └───────────────────┬────────────────────┘  │
         │                      ▼                       │
         │  ┌────────────────────────────────────────┐  │
         │  │ Encoder-Decoder Attention              │  │
         │  └───────────────────┬────────────────────┘  │
         │                      ▼                       │
         │  ┌────────────────────────────────────────┐  │
         │  │ Feed Forward Neural Network            │  │
         │  └───────────────────┬────────────────────┘  │
         └──────────────────────┼───────────────────────┘
                                ▼
                        Linear & Softmax
                                │
                                ▼
                       [ PALAVRA PREVISTA ]
```

###Processamento de Entrada (Preparando o texto)

Antes de entrar nos blocos principais, o texto passa por duas etapas:

• Input Embedding: Transforma as palavras (tokens) em vetores numéricos de alta dimensão que carregam o significado semântico inicial de cada palavra.

• Positional Encoding (Codificação Posicional): Como o Transformer processa todas as palavras ao mesmo tempo, ele perde a noção de ordem. Para resolver isso, ele soma vetores gerados por funções matemáticas (senos e cossenos) ao embedding original. Isso adiciona a informação da posição exata de cada palavra na frase.

###O Codificador (Encoder)

O artigo original empilha 6 blocos idênticos de codificação. O objetivo do Encoder é processar o texto de entrada e gerar uma representação abstrata rica em contexto. Cada bloco possui:

• Multi-Head Self-Attention (Atenção de Cabeças Múltiplas): Permite que cada palavra "olhe" para todas as outras palavras da frase simultaneamente para entender o contexto. Dividir isso em várias "cabeças" permite que o modelo aprenda diferentes tipos de relações ao mesmo tempo (ex: relações gramaticais, de sujeito-objeto, temporais).

• Feed-Forward Networks (FFN): Uma rede neural simples aplicada a cada posição de forma independente, adicionando não-linearidade e capacidade de aprendizado ao bloco.

• Conexões Residuais e Normalização de Camada (LayerNorm): Presentes após cada subcamada para garantir que o gradiente flua bem durante o treinamento, evitando que o modelo "esqueça" o que aprendeu no início.

### O Decodificador (Decoder)

O Decodificador também empilha 6 blocos idênticos. Ele é responsável por gerar a resposta (ou tradução) palavra por palavra. Cada bloco possui uma camada extra:

• Masked Multi-Head Self-Attention: Semelhante à atenção do Encoder, mas com uma máscara. Ela impede que a palavra atual "olhe para o futuro". Ao gerar a terceira palavra de uma frase, o modelo só pode olhar para a primeira e a segunda.

• Encoder-Decoder Attention: Esta camada conecta o Encoder ao Decoder. Ela permite que o Decoder consulte a representação final rica do texto de entrada gerada pelo Encoder, focando nas partes mais importantes para gerar a próxima palavra.

• Feed-Forward Networks: Idêntica à do Encoder.

### A Saída (A Palavra Final)

Após passar pelo último Decoder, os dados chegam à camada final:

• Linear & Softmax: A camada linear projeta a saída do modelo de volta para o tamanho do vocabulário do sistema. A função Softmax transforma esses valores em probabilidades. A palavra com a maior probabilidade é escolhida como o próximo elemento do texto gerado.

###Variações Modernas

Embora o design original use ambos (Encoder-Decoder), muitos modelos de sucesso hoje usam apenas partes dele:

• Apenas Encoder (ex: BERT): Excelente para entender o texto, classificar sentimentos ou fazer buscas.

• Apenas Decoder (ex: GPT, Gemini, Llama, Claude): Excelente para geração de texto e conversação.

Para entender como um LLM funciona internamente, vamos acompanhar o caminho percorrido por uma frase desde a entrada até a geração da resposta.

1\. Texto de entrada (prompt): "A inteligência artificial aprende"

2\. Tokenizador: Converte o texto em tokens e IDs numéricos

3\. Embeddings + informação posicional: Representa os tokens como vetores numéricos contextualizáveis

4\. Blocos Transformer repetidos:
- Multi-Head Self-Attention
- Identifica relações entre tokens
- Add & Norm
- Conexões residuais e normalização
- Feed-Forward Network (MLP)
- Transforma as representações aprendidas
- Add & Norm — o padrão se repete em várias camadas

5\. Camada de saída (LM Head): Calcula os escores e as probabilidades dos próximos tokens

6\. Seleção do próximo token: O token escolhido é acrescentado ao texto, e o processo continua

Essa é uma visão simplificada de um Transformer do tipo decoder-only, comum em modelos de geração de texto, como GPT e muitos outros LLMs. Existem outras arquiteturas, como modelos encoder-only e encoder-decoder.

### Tokenização (Tokenizer)

O modelo não recebe diretamente palavras como um ser humano. Primeiro, o texto é dividido em unidades chamadas tokens, que podem corresponder a palavras inteiras, partes de palavras, pontuação ou outros símbolos.

Exemplo ilustrativo:

Texto original: "**Os modelos aprendem padrões.**"

Possível divisão em tokens:
- Os
- modelos
- aprendem
- padrões
- .

A divisão real depende do tokenizador utilizado.

Cada token é convertido em um identificador numérico. O modelo trabalha com esses IDs para acessar suas representações vetoriais.

### Embeddings e informação posicional

Cada token é convertido em um vetor numérico de dimensão \\(d\_{\text{model}}\\), por meio de uma matriz de embeddings.

Por exemplo, um token pode ser representado por um vetor com milhares de valores, dependendo da arquitetura.

A informação posicional permite que o modelo considere a ordem dos tokens. Isso é importante porque:

- "O cão perseguiu o gato."
- "O gato perseguiu o cão."

As palavras são semelhantes, mas a ordem altera o significado.

Os Transformers podem usar embeddings posicionais aprendidos, codificações posicionais ou mecanismos como RoPE (Rotary Position Embedding), dependendo do modelo.

### Self-Attention: o mecanismo central

A *Self-Attention* permite que cada token considere outros tokens relevantes para interpretar o contexto.

Considere a frase:

> Maria entregou o livro a Ana porque ela precisava estudá-lo.

Para interpretar o pronome "ela", o modelo precisa considerar as relações entre as palavras da frase. O mecanismo de atenção ajuda a calcular essas relações, embora não garanta uma interpretação correta em todos os casos.

O mecanismo utiliza três representações derivadas dos vetores de entrada:

- Query (Q): o que o token procura no contexto.
- Key (K): quais características cada token oferece para ser relacionado.
- Value (V): a informação que cada token pode contribuir.

A expressão matemática fundamental da atenção é:

\\[ \operatorname{Attention}(Q,K,V)= \operatorname{softmax} \left(\frac{QK^\top}{\sqrt{d_k}}\right)V \\]

Em que:

- \\(QK^\top\\) calcula os escores de compatibilidade entre tokens.
- \\(d_k\\) é a dimensão dos vetores Key.
- `softmax` transforma os escores em pesos normalizados.
- A multiplicação por \\(V\\) combina as informações de acordo com esses pesos.

Nos LLMs generativos do tipo decoder-only, utiliza-se normalmente uma máscara causal: durante a geração, cada posição só pode acessar as posições anteriores e a própria posição, e não os tokens futuros.

### Multi-Head Attention

Em vez de utilizar apenas um mecanismo de *Self-Attention*, o Transformer pode utilizar várias cabeças (heads) em paralelo.

Cada cabeça aprende representações e relações diferentes. Depois, os resultados são combinados por uma transformação linear.

Isso permite que o modelo considere simultaneamente diferentes relações sintáticas, semânticas e contextuais.

### Feed-Forward Network (MLP)

Depois da *Self-Attention*, cada posição passa por uma rede neural do tipo feed-forward, também chamada MLP (Multi-Layer Perceptron).

A *Self-Attention* combina informações entre posições; a MLP transforma as representações de cada posição. Em conjunto, esses mecanismos permitem aprender padrões complexos.

### Conexões residuais e normalização

As conexões residuais (residual connections) permitem que informações de uma etapa sejam somadas à saída da transformação. A normalização (LayerNorm ou variantes, dependendo do modelo) ajuda a estabilizar o processamento e o treinamento.

Esses componentes são importantes quando se empilham dezenas ou centenas de blocos Transformer.

## Como os blocos se repetem?

Um LLM não possui apenas um bloco Transformer. Ele empilha muitos blocos, com parâmetros aprendidos durante o treinamento.

Bloco Transformer 1

*Self-Attention* + MLP + conexões residuais e normalização

Bloco Transformer 2

Refina as representações dos tokens

$\cdots$

Bloco Transformer N

Produz representações contextualizadas finais

LM Head + Softmax

Escores e probabilidades para o próximo token

A quantidade de blocos, a dimensão dos vetores, o número de cabeças e o tamanho das redes MLP variam conforme a arquitetura.

## Como o modelo gera uma resposta?

Ao terminar o processamento do prompt, um modelo autoregressivo calcula escores para os tokens possíveis do vocabulário. Uma função como `softmax` pode convertê-los em probabilidades.

Por exemplo, para o início de uma frase, o modelo poderia atribuir probabilidades ilustrativas:

| Próximo token possível | Probabilidade |
| ---------------------- | ------------- |
| aprender               | 45%           |
| analisar               | 30%           |
| processar              | 15%           |
| outros tokens          | 10%           |

O sistema seleciona um token segundo a estratégia de geração utilizada. O token é acrescentado à sequência, e o modelo calcula o próximo. O processo continua até atingir um token de parada ou outro critério definido.

Parâmetros como `temperature`, `top_k` e `top_p` podem influenciar a seleção dos tokens, quando suportados e ativados na configuração de geração.

## Como os LLMs são treinados?

A arquitetura é apenas uma parte do sistema. Os parâmetros do modelo precisam ser aprendidos por meio de treinamento.

Em um treinamento autoregressivo típico, o modelo recebe sequências de tokens e aprende a prever os tokens seguintes. Uma função de perda comum é a entropia cruzada (cross-entropy loss):

\\[ \mathcal{L}=-\sum_{t=1}^{T}\log P(x_t\mid x<{t}) \\]

Nessa expressão, \\(x_t\\) é o token esperado na posição \\(t\\), e \\(x{<t}\\) representa os tokens anteriores.

O treinamento ajusta os pesos da rede por meio de retropropagação (backpropagation) e algoritmos de otimização, como Adam ou AdamW.

Depois do pré-treinamento, alguns modelos passam por etapas adicionais, como ajuste supervisionado com exemplos de instruções e treinamento de preferência, para melhorar o comportamento conversacional.

## Outras arquiteturas de LLM

Nem todos os modelos de linguagem utilizam exatamente a mesma estrutura.

| Arquitetura              | Característica                                                           | Exemplos conhecidos                        |
| ------------------------ | ------------------------------------------------------------------------ | ------------------------------------------ |
| Encoder-only             | Representa e classifica textos considerando o contexto bidirecional      | BERT                                       |
| Decoder-only             | Prediz tokens seguintes e é muito usado para geração de texto            | GPT, famílias Llama e Qwen                 |
| Encoder-decoder          | Um encoder representa a entrada e um decoder produz a saída              | T5, BART                                   |
| Mixture of Experts (MoE) | Ativa seletivamente partes especializadas da rede em determinados tokens | Mixtral e variantes MoE de outras famílias |

Os exemplos são famílias de modelos, não garantias de que todas as suas versões tenham a mesma implementação interna. Além disso, MoE não é uma arquitetura concorrente ao Transformer: é uma estratégia que pode ser integrada aos blocos de um Transformer.

````python
# Inspecionando a arquitetura
# Com a biblioteca Transformers, pode-se visualizar a estrutura de um modelo real, como o GPT-2
from transformers import AutoModelForCausalLM
model = AutoModelForCausalLM.from_pretrained("openai-community/gpt2")
# Mostra a estrutura do modelo
print(model)
# Quantidade total de parâmetros
total = sum(p.numel() for p in model.parameters())
# Quantidade de parâmetros treináveis
treinaveis = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Parâmetros totais: {total:,}")
print(f"Parâmetros treináveis: {treinaveis:,}")
````

O código baixa os pesos do modelo se necessário. Para o GPT-2 original, espera-se aproximadamente 124 milhões de parâmetros na variante de cerca de 124M, embora a contagem exata dependa da configuração carregada.

A arquitetura interna de muitos LLMs modernos é construída em torno de embeddings, blocos Transformer com atenção e MLPs, e uma camada de saída que calcula os próximos tokens. A combinação desses componentes, seus parâmetros aprendidos e o processo de treinamento permite que o modelo processe contextos e gere textos complexos.

## Exemplo de utilização de LLM

O exemplo abaixo ilustra a ideia básica de enviar uma pergunta a um modelo e receber uma resposta utilizando a biblioteca Transformers, do ecossistema Hugging Face.

Como funciona?

1. `pipeline()` configura uma tarefa de geração de texto.
2. `model="gpt2"` seleciona um modelo pré-treinado.
3. `prompt` fornece o texto inicial que orienta a geração.
4. `max_new_tokens=50` limita a geração a, no máximo, 50 novos tokens.
5. `generated_text` contém o texto produzido pelo modelo.

Observação: o GPT-2 é um modelo introdutório relativamente pequeno, útil para demonstrações. Sua capacidade de seguir instruções e responder perguntas é inferior à de muitos LLMs modernos.

Em resumo, Hugging Face facilita o acesso aos modelos, Ollama facilita sua execução local e LangChain facilita a construção de aplicações que utilizam esses modelos. Essas ferramentas podem ser utilizadas separadamente ou combinadas, dependendo dos objetivos do projeto.

````python
from transformers import pipeline, GenerationConfig

gerador = pipeline("text-generation",  model="gpt2")
# Define uma configuração explícita de geração
config = GenerationConfig(max_new_tokens=100, do_sample=True)
````

````python
import textwrap

def gpt2(prompt, gerador, config):
  prompt = prompt
  gerador = gerador
  config = config
  resultado = gerador(prompt, generation_config=config, clean_up_tokenization_spaces=False)
  texto = resultado[0]["generated_text"]
  print(textwrap.fill(texto, width=100))
````

````python
prompt = "Artificial Intelligence can help"
prompt = "Machine learning is useful because "
prompt = "Renewable energy can contribute to"
prompt = "Deep learning is a technique that"
prompt = "I can't believe you did such a "
# prompt = "I can get big in AI"
# prompt = "Charles Chaplin was"
gpt2(prompt, gerador, config)
````

## Biblioteca Transformers da Hugging Face

A Hugging Face é uma plataforma que disponibiliza um amplo repositório de modelos de Inteligência Artificial pré-treinados, incluindo modelos de linguagem de código aberto ou com pesos publicamente disponíveis, prontos para serem carregados e utilizados em tarefas de Processamento de Linguagem Natural (PLN), como geração de texto, tradução automática e análise de sentimentos.

Seu principal componente é a biblioteca Transformers, que oferece integração com bibliotecas populares do Python, como PyTorch, JAX e TensorFlow, facilitando a implementação e a utilização de modelos de aprendizado profundo.

Uma das principais vantagens dessa plataforma é simplificar o desenvolvimento de aplicações de IA. Muitos modelos podem ser utilizados gratuitamente e exigem poucas configurações iniciais, tornando o desenvolvimento de soluções baseadas em Inteligência Artificial mais acessível.

Entendendo os principais comandos:

- `pipeline()`: simplifica a utilização de modelos para tarefas específicas.
- `"text-generation"`: indica que a tarefa é geração de texto.
- `model="openai-community/gpt2"`: identifica o modelo a ser carregado do Hugging Face Hub.
- `prompt`: fornece o texto inicial para orientar a geração.
- `max_new_tokens=50`: permite gerar até 50 tokens adicionais.
- `do_sample=True`: habilita a amostragem, permitindo variar as continuações produzidas.

Depois de executar o exemplo, altere o prompt para observar como o texto gerado muda:

- `"Machine learning is useful because"`
- `"Renewable energy can contribute to"`
- `"Deep learning is a technique that"`

Lembre-se de que o GPT-2 é um modelo introdutório de geração de texto. Ele pode produzir continuações incoerentes ou informações incorretas; portanto, os resultados servem principalmente para compreender o funcionamento da biblioteca.

Para aprofundar seus estudos, consulte a documentação oficial: [Hugging Face Transformers](https://huggingface.co/docs/transformers/).

Exemplo da utilização do modelo do Hugging Face — especificamente o GPT-2 — para a geração de texto. Ao carregar modelos pré-treinados, geralmente é necessário importar não apenas o modelo em si, mas também um tokenizador compatível. Esse componente é responsável por dividir os textos de entrada em unidades linguísticas lógicas chamadas "tokens" (que, na maioria dos casos, equivalem aproximadamente a palavras) antes que sejam enviados ao modelo de linguagem para processamento e geração de texto complementar.

O código a seguir importa os pacotes necessários e inicializa o modelo de linguagem e o tokenizador.

Em seguida, definimos um estímulo para o modelo, que normalmente é uma consulta, pergunta ou solicitação para a qual o modelo de linguagem tentará gerar uma resposta. Alguns modelos menos sofisticados se limitam a gerar palavras subsequentes que continuam a sequência de entrada ou o estímulo, por exemplo, para continuar uma história que começa com "Era uma vez". No nosso caso, tentaremos perguntar ao GPT-2, qual é a capital do Japão?

````python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

model_name = "gpt2"
model = GPT2LMHeadModel.from_pretrained(model_name)
tokenizer = GPT2Tokenizer.from_pretrained(model_name)

prompt = "What is the capital of Japan?"
# prompt = "Artificial Intelligence can help us?"
# prompt = "What is deep learning used for?"

inputs = tokenizer.encode(prompt, return_tensors="pt")
output = model.generate(inputs, max_length=100, num_return_sequences=1)

response = tokenizer.decode(output[0], skip_special_tokens=True)
print(textwrap.fill(response, width=100))
````

Entrando em detalhes técnicos, no código acima você talvez tenha notado que ocorrem duas etapas de processamento: codificação e decodificação. A sequência de entrada (o prompt) precisa ser codificada em uma representação vetorial numérica que o modelo consiga compreender; após o modelo gerar a resposta bruta (numérica), ela é decodificada de volta para texto, tornando-a compreensível para nós.

Bem, pelo menos ele deu a resposta certa! Esses modelos pré-treinados são comparativamente menores, mais fáceis de manejar em ambientes de teste e não são tão absurdamente poderosos ou de alto desempenho quanto os modelos do ChatGPT; portanto, não é de se surpreender que apresentem limitações maiores na geração de texto. Em particular, como definimos um comprimento máximo de 100 tokens para a resposta, o modelo parece priorizar a entrega de um conteúdo que se aproxime desse limite, em vez de focar em uma resposta concisa, simples e lógica.

### O fluxo essencial (Tokenização → Modelo → Decodificação)
O exemplo de código abaixo mostra o "coração" de uma LLM. Vamos usar o GPT-2, que é pequeno e didático.

````python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# 1. Carregar o tokenizador e o modelo
# O tokenizador converte texto em números (IDs) que o modelo entende.
# O modelo contém os pesos treinados (a "inteligência").
model_id = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)

# 2. Tokenizar uma frase de entrada (prompt)
# return_tensors="pt" diz que queremos tensores do PyTorch.
prompt = "Machine learning is useful because"
inputs = tokenizer(prompt, return_tensors="pt")

print("IDs dos tokens:", inputs["input_ids"])
# Saída esperada (exemplo): tensor([[ 32, 12045, 16, 11477, 13]])
````

````python
inputs['attention_mask']
````

````python
# 3. Passar pelo modelo (forward pass)
# O modelo recebe os IDs e cospe "logits" (pontuações brutas para cada palavra do vocabulário)
with torch.no_grad():  # Desliga o cálculo de gradientes (só queremos inferência)
    outputs = model(**inputs)

logits = outputs.logits
print("Formato dos logits:", logits.shape)
````

````python
model(**inputs)
````

````python
# 4. Pegar a predição para o PRÓXIMO token
# O modelo prevê o próximo token baseado no ÚLTIMO token da sequência.
# logits[0, -1, :] pega as pontuações para a última posição.
proximos_logits = logits[0, -1, :]
proximo_token_id = torch.argmax(proximos_logits).item()  # Pega o ID com maior pontuação

print("ID do próximo token:", proximo_token_id)
````

````python
# 5. Decodificar o ID de volta para texto
proximo_token = tokenizer.decode(proximo_token_id)
print(f"Prompt: '{prompt}'")
print(f"Próximo token previsto: '{proximo_token}'")
# Saída provável: "Próximo token previsto: ' it'" (ou similar)
````

### O que aconteceu?
O modelo não "pensou" em uma frase inteira. Ele apenas calculou: "Dada a sequência 'Machine learning is useful because', qual é a palavra mais provável a seguir?" . Em um chat real, esse processo é repetido em um loop: o token previsto é anexado à entrada, e o modelo prevê o próximo, até formar uma resposta completa.

### Gerando texto de forma automática
Fazer o loop manualmente é inviável. A função `generate` faz isso para nós:

````python
# Usando o mesmo modelo e tokenizador do exemplo anterior

prompt = "Brazil is a country where"
# prompt = "Machine learning is useful because"
prompt = "What is the capital of Japan?"
inputs = tokenizer(prompt, return_tensors="pt")

# generate() cuida do loop de predição automática
outputs = model.generate(
    inputs["input_ids"],
    # max_new_tokens: Limite de tokens o modelo pode adicionar.
    max_new_tokens=20,
    # do_sample=True: Ativa amostragem (torna o texto menos repetitivo)
    do_sample=True,
    # temperature: quanto menor (ex: 0.2), mais "seguro" e repetitivo;
    # quanto maior (ex: 1.0), mais aleatório e criativo .
    temperature=0.7, # 0.1,
    # top_k: em vez de considerar todas as 50 mil palavras, considera apenas as k mais prováveis.
    # Isso evita que o modelo escolha palavras absurdas
    top_k=50
)
# convert tokens to string
texto_gerado = tokenizer.decode(outputs[0], skip_special_tokens=True)
print('Tokens:\n', outputs[0])
print('\nTokens convertidos em texto:\n', texto_gerado)
````

Vamos começar com os LLMs (Large Language Models) e os transformers.

**LLMs (Large Language Models)**

Os LLMs são um tipo de modelo de linguagem treinado com grandes quantidades de dados. Eles são projetados para processar e gerar texto de forma autônoma, sem a necessidade de uma interface de usuário.

Os LLMs são treinados com grandes conjuntos de dados, como livros, artigos e textos de internet. Isso permite que eles aprendam a estrutura e o padrão da linguagem humana, incluindo a gramática, a sintaxe e a semântica.

Os LLMs são usados em uma variedade de aplicações, como:

*   **Resposta automática**: os LLMs podem responder a perguntas e fornecer informações baseadas nos dados treinados.
*   **Geração de texto**: os LLMs podem gerar texto de forma autônoma, como artigos, resumos e até mesmo histórias.
*   **Tradução**: os LLMs podem traduzir texto de um idioma para outro.

**Transformers**

O transformer é um tipo de arquitetura de rede neural que é amplamente usada em modelos de linguagem, incluindo os LLMs.

A ideia fundamental do transformer é que a atenção (ou "atenção mútua") é uma operação que permite que o modelo "ouça" e "entenda" o contexto em que está sendo usado.

Em outras palavras, o transformer permite que o modelo considere todas as partes de um texto ao mesmo tempo, em vez de processá-lo de forma sequencial.

Isso permite que os modelos de linguagem sejam mais eficientes e precisos, pois podem capturar relações mais complexas entre as palavras e as ideias.

**Exemplo simples**

Imagine que você está tentando escrever uma história. Você pode começar com uma frase, como "O sol estava a pôr de lado". Em vez de escrever apenas essa frase, você pode usar um modelo de linguagem treinado com transformers para gerar a próxima frase, como "e iluminava a praia de areia branca".

O modelo de linguagem está "ouvindo" a frase anterior e "entendendo" o contexto, e então gera a próxima frase com base nisso.

Essa é a ideia básica dos transformers e LLMs. Eles são ferramentas poderosas que permitem que os modelos de linguagem sejam mais eficientes e precisos, e que possam processar e gerar texto de forma autônoma.
